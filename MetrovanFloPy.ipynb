```python
import numpy as np, shutil, os

# Replace these with the real min-X, max-Y from your own model_domain layer's extent
X0_van, Y0_van = 1210000.0, 480000.0   # <-- put your real domain bounds here

nrow_v, ncol_v = 40, 40
cell_size_v = 100

top_v = 80.0
botm_v = [top_v - 30]
k_vancouver = 15.0
recharge_vancouver = 0.0009
strt_v = top_v - 8

model_dir_v = "/content/vancouver_model"
if os.path.exists(model_dir_v):
    shutil.rmtree(model_dir_v)
print("Vancouver setup ready.")
```

```python
import flopy
sim_v = flopy.mf6.MFSimulation(sim_name="vancouver_flow", sim_ws=model_dir_v)
flopy.mf6.ModflowTdis(sim_v, time_units="DAYS", nper=1, perioddata=[(1.0, 1, 1.0)])
flopy.mf6.ModflowIms(sim_v)
gwf_v = flopy.mf6.ModflowGwf(sim_v, modelname="vancouver_flow", save_flows=True)
flopy.mf6.ModflowGwfdis(gwf_v, nlay=1, nrow=nrow_v, ncol=ncol_v, delr=cell_size_v, delc=cell_size_v, top=top_v, botm=botm_v)
flopy.mf6.ModflowGwfic(gwf_v, strt=strt_v)
print("Container built.")
```

```python
flopy.mf6.ModflowGwfnpf(gwf_v, icelltype=1, k=k_vancouver)
flopy.mf6.ModflowGwfrcha(gwf_v, recharge=recharge_vancouver)
print("K and recharge set.")
```

```python
chd_data_v = (
    [[(0, r, 0), 75.0] for r in range(nrow_v)] +
    [[(0, r, ncol_v - 1), 65.0] for r in range(nrow_v)]
)
flopy.mf6.ModflowGwfchd(gwf_v, stress_period_data=chd_data_v)
flopy.mf6.ModflowGwfwel(gwf_v, stress_period_data={0: [[(0, nrow_v // 2, ncol_v // 2), -300.0]]})
flopy.mf6.ModflowGwfoc(gwf_v, head_filerecord="vancouver_flow.hds", saverecord=[("HEAD", "ALL")])
print("Boundaries and well set.")
```

```python
sim_v.write_simulation()
success_v, buffer_v = sim_v.run_simulation()
print("Vancouver model ran successfully:", success_v)
if not success_v:
    for line in buffer_v: print(line)
```

```python
import flopy.utils as futil
import matplotlib.pyplot as plt
import numpy as np

heads_v = futil.HeadFile(f"{model_dir_v}/vancouver_flow.hds").get_data()
masked_v = np.ma.masked_where(heads_v[0] <= -1e29, heads_v[0])
print("Min:", masked_v.min(), "Max:", masked_v.max())

fig, ax = plt.subplots()
im = ax.imshow(masked_v, cmap="Blues")
fig.colorbar(im, ax=ax, label="Water level (m)")
ax.set_title("Metro Vancouver — simulated water table")
plt.show()
```

```python
import rasterio
from rasterio.transform import from_origin

top_left_x_v = X0_van
top_left_y_v = Y0_van + nrow_v * cell_size_v
transform_v = from_origin(top_left_x_v, top_left_y_v, cell_size_v, cell_size_v)

with rasterio.open("vancouver_heads.tif", "w", driver="GTiff",
                    height=nrow_v, width=ncol_v, count=1,
                    dtype=masked_v.dtype, crs="EPSG:3005", transform=transform_v) as dst:
    dst.write(masked_v.filled(-9999), 1)

from google.colab import files
files.download("vancouver_heads.tif")
```
