```python
import numpy as np
import shutil, os

model_dir = "/content/kamloops_model"
if os.path.exists(model_dir):
    shutil.rmtree(model_dir)

X0, Y0 = 1389861.0, 641869.0
nrow_k, ncol_k = 30, 30
cell_size_k = 20

top_k = 380.0
botm_k = [top_k - 40]
recharge_kamloops = 0.00004
strt_k = top_k - 15

print("Grid and values ready:", nrow_k, "x", ncol_k, "cells, top =", top_k)
```

```python
sim_k = flopy.mf6.MFSimulation(sim_name="kamloops_flow", sim_ws=model_dir)
flopy.mf6.ModflowTdis(sim_k, time_units="DAYS", nper=1, perioddata=[(1.0, 1, 1.0)])
flopy.mf6.ModflowIms(sim_k)
gwf_k = flopy.mf6.ModflowGwf(sim_k, modelname="kamloops_flow", save_flows=True)

flopy.mf6.ModflowGwfdis(gwf_k, nlay=1, nrow=nrow_k, ncol=ncol_k,
                         delr=cell_size_k, delc=cell_size_k, top=top_k, botm=botm_k)
flopy.mf6.ModflowGwfic(gwf_k, strt=strt_k)
print("Model container and grid built.")
```

```python
k_array = np.full((nrow_k, ncol_k), 0.5)
k_array[20:30, :] = 25.0
flopy.mf6.ModflowGwfnpf(gwf_k, icelltype=1, k=k_array)
flopy.mf6.ModflowGwfrcha(gwf_k, recharge=recharge_kamloops)
print("K zones and recharge set.")
```

```python
chd_data_k = (
    [[(0, 0, c), 375.0] for c in range(ncol_k)] +
    [[(0, nrow_k - 1, c), 350.0] for c in range(ncol_k)]
)
flopy.mf6.ModflowGwfchd(gwf_k, stress_period_data=chd_data_k)

flopy.mf6.ModflowGwfwel(gwf_k, stress_period_data={0: [
    [(0, 8, 12), -800.0],
    [(0, 8, 18), -800.0],
]})

flopy.mf6.ModflowGwfoc(gwf_k, head_filerecord="kamloops_flow.hds", saverecord=[("HEAD", "ALL")])
print("Boundaries and wells set.")
```

```python
sim_k.write_simulation()
success_k, buffer_k = sim_k.run_simulation()
print("Kamloops model ran successfully:", success_k)

if not success_k:
    print("Error log:")
    for line in buffer_k:
        print(line)
```

```python
import matplotlib.pyplot as plt
import flopy.utils as futil

heads_k = futil.HeadFile(f"{model_dir}/kamloops_flow.hds").get_data()
fig, ax = plt.subplots()
im = ax.imshow(heads_k[0], cmap="Blues")
fig.colorbar(im, ax=ax, label="Water level (m)")
ax.set_title("Kamloops mine site — simulated water table (two K zones)")
plt.show()
```
